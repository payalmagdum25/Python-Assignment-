```python
%sql
--Cell 30 – Find books with their authors (JOIN)
SELECT
    b.book_id,
    b.title,
    a.author_name,
    b.category
FROM library_mgmt.Books b

INNER JOIN library_mgmt.Book_Authors ba
    ON b.book_id = ba.book_id

INNER JOIN library_mgmt.Authors a
    ON ba.author_id = a.author_id;
```

```python
%sql
--Cell 31 – Find available books
SELECT
    b.book_id,
    b.title,
    COUNT(c.copy_id) AS available_copies
FROM library_mgmt.Books b

INNER JOIN library_mgmt.Book_Copies c
    ON b.book_id = c.book_id

WHERE c.copy_status = 'AVAILABLE'

GROUP BY
    b.book_id,
    b.title

ORDER BY available_copies DESC;
```

```python
%sql
--Find overdue books 
SELECT
    l.loan_id,
    m.full_name,
    b.title,
    l.issue_date,
    l.due_date,
    l.return_date
FROM library_mgmt.Loans l
INNER JOIN library_mgmt.Members m
    ON l.member_id = m.member_id
INNER JOIN library_mgmt.Book_Copies c
    ON l.copy_id = c.copy_id
INNER JOIN library_mgmt.Books b
    ON c.book_id = b.book_id
WHERE
    l.return_date IS NULL
    AND l.due_date < CURRENT_DATE();
```

```python
%sql
--calculate the overdue fines
SELECT
    l.loan_id,
    m.full_name,
    b.title,
    l.due_date,
    DATEDIFF(CURRENT_DATE(), l.due_date) AS overdue_days,
    DATEDIFF(CURRENT_DATE(), l.due_date) * 10 AS calculated_fine
FROM library_mgmt.Loans l
INNER JOIN library_mgmt.Members m
    ON l.member_id = m.member_id
INNER JOIN library_mgmt.Book_Copies c
    ON l.copy_id = c.copy_id
INNER JOIN library_mgmt.Books b
    ON c.book_id = b.book_id
WHERE
    l.return_date IS NULL
    AND l.due_date < CURRENT_DATE();
```

```python
# CELL 34: Transformations and Actions

# Read the Books table
books_df = spark.table("library_mgmt.Books")

# TRANSFORMATIONS
# These operations build the execution plan
programming_books = books_df.filter(
    books_df.category == "Programming"
)

selected_books = programming_books.select(
    "book_id",
    "title",
    "publication_year"
)

# ACTION
# show() triggers execution
selected_books.show()
```

```python
# CELL 35: View Spark Execution Plan (DAG)

selected_books.explain(mode="formatted")
```

```python
# CELL 36: Demonstrate Spark Actions

# Action 1: Count total books
total_books = books_df.count()

print("Total Books:", total_books)

# Action 2: Display all books
books_df.show()

print("Both actions executed successfully!")
```

```python
# CELL 37: Demonstrate Partitioning

# Read Books table
books_df = spark.table("library_mgmt.Books")

# Create 4 partitions
partitioned_books = books_df.repartition(4)

# Display partitioning execution plan
partitioned_books.explain(mode="formatted")

# Display records
partitioned_books.show()
```

```python
# CELL 38: Demonstrate Shuffle

# Group books by category and count them
category_count = (
    books_df
    .groupBy("category")
    .count()
)

# Display execution plan
category_count.explain(mode="formatted")

# Execute the query
category_count.show()
```

```python
# CELL 39: Create a Parquet Table

# Read existing Books table
books_df = spark.table("library_mgmt.Books")

# Save data in Parquet format
books_df.write \
    .format("parquet") \
    .mode("overwrite") \
    .saveAsTable("library_mgmt.Books_Parquet")

print("Parquet table created successfully!")
```
